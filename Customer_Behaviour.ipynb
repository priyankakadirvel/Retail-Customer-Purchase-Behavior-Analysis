Mean - outliers
overall median is wrong as it creates biased

Instead median is found for each category and replace accordingly



```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

SNAKE CASING the column names

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.head(2)
```

```python
df =  df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group

labels = ['Young Adult','Adult',' Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = labels) #4 equal size grp
```

```python
df[['age','age_group']].head()
```

```python
#create column purchase_frequency_days

frequency_mapping = {'Fortnightly' : 14,
                     'Weekly' : 7,
                     'Monthly' : 30,
                     'Quarterly' : 90,
                     'Bi-Weekly' : 14,
                     'Annually': 365,
                     'Every 3 Months' : 90
                     }

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(5)
```

```python
df[['discount_applied','promo_code_used']].head()
```

```python
df['discount_applied'] = df['promo_code_used'].all()
```

```python
df = df.drop('promo_code_used',axis = 1)
```
